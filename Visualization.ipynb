```python
## Lukas's Code

import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('dark_background') ## Accessibility

pd.set_option('display.max_columns', None)
pd.set_option('display.width',1000)


# CSVs converted into DataFrame
mycsv = pd.read_csv('C:\\Users\\lukas\\Files\\Dev\\Globe-At-Night\\Globe-At-Night-Visualization\\GlobeAtNight2025.csv')
#print("CSV")
#print(mycsv)

# Removed the null values in the dataset, Preprocessing
mycsv.dropna(subset=["LimitingMag"], inplace=True)
mycsv.dropna(subset=["SQMReading"], inplace=True)
mycsv.index=mycsv.index.astype(int)
#print(copy_mycsv)

# Start LimitingMag graph

plt.figure(figsize=(6,4))

#experimenting to see what happens when markersize is changed, helps me see better due to my vision issues

sns.histplot(data=mycsv, x='LimitingMag', bins=20, kde=False, color='green')

plt.title("LimitingMag WorldWide")


plt.xlabel("LimitingMag")
plt.ylabel("Frequency")


plt.legend()


#plt.grid(True)


#plt.tight_layout()


plt.show()
plt.close()


print("LimitingMag max =",mycsv["LimitingMag"].max())
print("LimitingMag min =",mycsv["LimitingMag"].min())
magcsv = mycsv['LimitingMag']
print("The mean Magnitude category across the entire dataset (measured with categories 0-7) is", magcsv.mean())

```

```python
countrycsv= mycsv['Country']
#print(countrycsv)

# Check row-by-row if the value exists anywhere in the row
ncdata = mycsv[mycsv.apply(lambda row: 'United States - North Carolina' in row.values, axis=1)]
#print(ncdata)
#print(mycsv)



# Start LimitingMag for NC graph

plt.figure(figsize=(6,4))

#experimenting to see what happens when markersize is changed, helps me see better due to my vision issues

sns.histplot(data=ncdata, x='LimitingMag', bins=20, kde=False, color='green')

plt.title("LimitingMag NC")


plt.xlabel("LimitingMag")
plt.ylabel("Frequency")


plt.legend()


#plt.grid(True)


#plt.tight_layout()


plt.show()
plt.close()

print("The mean magnitude category across just North Carolina (measured with categories 0-7) is",ncdata['LimitingMag'].mean())
print("However, the shape of the graph is pulling the mean to the right.")
```
