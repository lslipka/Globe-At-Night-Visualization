```python
## Lukas's Code

import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
pd.set_option('display.width',1000)


# CSVs converted into DataFrame
mycsv = pd.read_csv('C:\\Users\\lukas\\Files\\Dev\\Globe-At-Night\\Globe-At-Night-Visualization\\GlobeAtNight2025.csv')
#print("CSV")
#print(mycsv)

magcsv = mycsv['LimitingMag']
print("The mean Magnitude category across the entire dataset (measured with categories 0-7) is", magcsv.mean())

# Removed the null values in the dataset
mycsv.dropna(subset=["LimitingMag"], inplace=True)
mycsv.dropna(subset=["SQMReading"], inplace=True)
mycsv.index=mycsv.index.astype(int)
#print(copy_mycsv)

# Start LimitingMag graph

plt.figure(figsize=(6,4))

#experimenting to see what happens when markersize is changed, helps me see better due to my vision issues

sns.histplot(data=mycsv, x='LimitingMag', bins=20, kde=False, color='green')

plt.title("LimitingMag WorldWide - Histogram")


plt.xlabel("LimitingMag")
plt.ylabel("Frequency")


plt.legend()


#plt.grid(True)


#plt.tight_layout()


plt.show()
plt.close()


print("LimitingMag max =",mycsv["LimitingMag"].max())
print("LimitingMag min =",mycsv["LimitingMag"].min())
```
